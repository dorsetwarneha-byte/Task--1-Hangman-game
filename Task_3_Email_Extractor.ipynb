```python
import re

with open("input.txt", "r") as file:
    text = file.read()

emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', text)

print("Emails found:")
for email in emails:
    print(email)
```

```python
with open("emails.txt", "w") as file:
    for email in emails:
        file.write(email + "\n")

print("Emails saved successfully!")
```

```python
with open("emails.txt", "r") as file:
    content = file.read()

print(content)
```
