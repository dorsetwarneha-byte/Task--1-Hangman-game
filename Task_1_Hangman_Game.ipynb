# 🎮 Hangman Game

### Python Programming Internship — Task 1

A simple text-based Hangman game developed using Python.

**Features:**
- Random word selection
- Letter-by-letter guessing
- Maximum 6 incorrect guesses
- Repeated guess detection
- Win and lose conditions

```python
import random

words = ["python", "computer", "analytics", "database", "programming"]

print("Words loaded successfully!")
```

```python
word = random.choice(words)

print("A word has been selected!")
print("Number of letters:", len(word))
```

```python
hidden_word = ["_"] * len(word)

print(" ".join(hidden_word))
```

```python
guessed_letters = []
wrong_guesses = 0
max_wrong_guesses = 6

print("You have", max_wrong_guesses, "wrong guesses available.")
```

```python
guess = input("Guess a letter: ").lower()

print("You guessed:", guess)
```

```python
if guess in word:
    print("Correct guess! 🎉")
else:
    wrong_guesses += 1
    print("Wrong guess! ❌")
    print("Wrong guesses:", wrong_guesses)
```

```python
for i in range(len(word)):
    if word[i] == guess:
        hidden_word[i] = guess

print(" ".join(hidden_word))
```

```python
if guess not in guessed_letters:
    guessed_letters.append(guess)

print("Guessed letters:", guessed_letters)
```

```python
if guess in guessed_letters:
    print("You already guessed this letter!")
else:
    guessed_letters.append(guess)
    print("Guessed letters:", guessed_letters)
```

```python
while wrong_guesses < max_wrong_guesses and "_" in hidden_word:
    guess = input("Guess a letter: ").lower()

    if guess in guessed_letters:
        print("You already guessed this letter!")
        continue

    guessed_letters.append(guess)

    if guess in word:
        print("Correct guess! 🎉")

        for i in range(len(word)):
            if word[i] == guess:
                hidden_word[i] = guess
    else:
        wrong_guesses += 1
        print("Wrong guess! ❌")
        print("Wrong guesses:", wrong_guesses)

    print("Word:", " ".join(hidden_word))
    print("Guessed letters:", guessed_letters)
```

```python
if "_" not in hidden_word:
    print("🎉 Congratulations! You won!")
    print("The word was:", word)
else:
    print("😢 Game Over!")
    print("The word was:", word)
```
