# Stock Portfolio Tracker

Python Programming Internship - Task 2

This project calculates stock investment values based on
user-entered stock names and quantities.

stock_prices = {
    'AAPL': 180,
    'TSLA': 250,
    'GOOGLE': 140,
    'MSFT': 320,
    'AMZN': 130
}

print(stock_prices)

```python
total_investment = 0

for i in range(3):
    stock = input('enter stock name:').upper()
    quantity = int (input('enter quantity:'))

    if stock in stock_prices:
        price = stock_prices[stock]
        investment = price * quantity

        print('stock:', stock)
        print('price:', price)
        print('quantity:',quantity)
        print('investment:', investment)

        total_investment = total_investment + investment
    else:
        print('stock not found!')

print('total investment:',total_investment)        
```

```python
print("----- STOCK PORTFOLIO SUMMARY -----")
print("Total Investment:", total_investment)
print("Number of Stocks:", 3)
print("-----------------------------------")
```

```python
with open("stock_portfolio.txt", "w") as file:
    file.write("STOCK PORTFOLIO SUMMARY\n")
    file.write("-----------------------\n")
    file.write("Total Investment: " + str(total_investment) + "\n")

print("Portfolio saved successfully!")
```

```python
with open("stock_portfolio.txt", "r") as file:
    content = file.read()

print(content)
```
